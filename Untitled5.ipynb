```python
import pandas as pd

# 1. Chargement standard (Pandas convertit automatiquement les cases vides en NaN)
df = pd.read_csv("projet.csv")

# 2. Liste des chaînes qui représentent de "faux positifs" textuels
fake_nulls = ["-", "N/A", "n/a", "null", "NULL", "NaN", "nan", "Rs.N/A"]

for col in df.columns:
    # Vrais NaN détectés par Pandas (cases vides du CSV)
    vrais_nulls = df[col].isna().sum()

    # Faux nulls (valeurs textuelles piégées qu'il faut nettoyer)
    valeurs_text = df[col].dropna().astype(str).str.strip()
    faux_nulls = valeurs_text.isin(fake_nulls).sum()

    print(f"{col}:")
    print(f"  Vrais NaN   : {vrais_nulls}")
    print(f"  Faux nulls  : {faux_nulls}")
```

```python
import pandas as pd
import re

# 1. Expression régulière adaptée aux Roupies Indiennes (Rs.)
pattern = r"^Rs\.\d{1,3}(,\d{3})*$"

# 2. Application du masque de conformité
conformes = df["Price"].astype(str).str.strip().str.match(pattern, na=False)

# 3. Calcul et affichage des résultats
taux_conformite = conformes.mean() * 100

print(f"Taux de conformité : {taux_conformite:.2f} %")

if taux_conformite < 100:
    print("\nValeurs non conformes :")
    print(df.loc[~conformes, "Price"].value_counts())
```

```python
import pandas as pd
import re

# Charger le dataset
df = pd.read_csv("projet.csv")


# ==========================================================
# FONCTION : analyser les schémas d'une colonne
# ==========================================================

def analyser_schemas(df, colonne):

    print("\n" + "=" * 70)
    print("COLONNE :", colonne)
    print("=" * 70)

    # Supprimer les valeurs nulles et les espaces inutiles
    serie = df[colonne].dropna().astype(str).str.strip()

    # ------------------------------------------------------
    # 1. Nombre total de valeurs
    # ------------------------------------------------------

    total = len(serie)

    print("\nNombre total de valeurs :", total)

    # ------------------------------------------------------
    # 2. Valeurs les plus fréquentes
    # ------------------------------------------------------

    print("\nPrincipales valeurs :")
    print(serie.value_counts().head(15))

    # ------------------------------------------------------
    # 3. Identifier les grands schémas
    # ------------------------------------------------------

    def schema(valeur):

        # Per_Dis_Cam_Bat
        if colonne == "Per_Dis_Cam_Bat":

            contient_display = "Display" in valeur
            contient_camera = "Camera" in valeur
            contient_battery = "Battery" in valeur
            contient_performance = "Performance" in valeur

            if contient_performance and contient_display and contient_camera and contient_battery:
                return "Performance + Display + Camera + Battery"

            elif contient_display and contient_camera and contient_battery:
                return "Display + Camera + Battery"

            elif contient_display and contient_camera:
                return "Display + Camera"

            elif contient_camera and contient_battery:
                return "Camera + Battery"

            else:
                return "Autre"

        # Other_Features
        elif colonne == "Other_Features":

            contient_sim = "SIM" in valeur
            contient_india = "Supported in India" in valeur or "Not Supported in India" in valeur
            contient_fm = "FM Radio" in valeur
            contient_fingerprint = "Fingerprint sensor" in valeur
            contient_otg = "USB OTG Support" in valeur
            contient_expandable = "Expandable" in valeur

            if contient_sim and contient_india and contient_fm:
                return "SIM + India + FM Radio"

            elif contient_sim and contient_india and contient_expandable:
                return "Storage + SIM + India"

            elif contient_sim and contient_india:
                return "SIM + India"

            elif contient_expandable:
                return "Storage / Expandable"

            elif contient_fingerprint or contient_otg:
                return "Security / Connectivity"

            else:
                return "Autre"

    # ------------------------------------------------------
    # 4. Appliquer la classification
    # ------------------------------------------------------

    schemas = serie.apply(schema)

    # ------------------------------------------------------
    # 5. Compter les schémas
    # ------------------------------------------------------

    repartition = schemas.value_counts()

    print("\nPrincipaux schémas :")
    print(repartition)

    # ------------------------------------------------------
    # 6. Calculer le taux de chaque schéma
    # ------------------------------------------------------

    print("\nTaux par schéma :")

    for nom_schema, nombre in repartition.items():

        taux = (nombre / total) * 100

        print(
            f"{nom_schema} : "
            f"{nombre} valeurs "
            f"({taux:.2f} %)"
        )

    # ------------------------------------------------------
    # 7. Valeurs marginales
    # ------------------------------------------------------

    marginales = serie[schemas == "Autre"]

    nombre_marginal = len(marginales)

    taux_marginal = (nombre_marginal / total) * 100

    print("\nValeurs marginales :", nombre_marginal)
    print("Taux de valeurs marginales :", round(taux_marginal, 2), "%")

    if nombre_marginal > 0:

        print("\nExemples de valeurs marginales :")
        print(marginales.value_counts().head(10))

    else:

        print("\nAucune valeur marginale.")


# ==========================================================
# ANALYSE DE Per_Dis_Cam_Bat
# ==========================================================

analyser_schemas(
    df,
    "Per_Dis_Cam_Bat"
)


# ==========================================================
# ANALYSE DE Other_Features
# ==========================================================

analyser_schemas(
    df,
    "Other_Features"
)
```
